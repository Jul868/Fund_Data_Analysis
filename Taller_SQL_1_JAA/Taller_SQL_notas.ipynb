```python
#ROW_NUMBER
query = """
SELECT
    v.venta_id,
    e.nombre AS vendedor,
    v.producto,
    v.cantidad * v.precio_unit AS monto,
    ROW_NUMBER() OVER (
        PARTITION BY e.nombre
        ORDER BY v.cantidad * v.precio_unit DESC
    ) AS num_venta
FROM ventas v
INNER JOIN empleados e ON e.empleado_id = v.empleado_id
"""
pd.read_sql_query(query, conn)
```

```python
#RANK
query = """
SELECT
    v.venta_id,
    v.categoria,
    v.cantidad * v.precio_unit AS monto,
    RANK() OVER (
        PARTITION BY v.categoria
        ORDER BY v.cantidad * v.precio_unit DESC
    ) AS ranking
FROM ventas v
"""
pd.read_sql_query(query, conn)
```

```python
#SUMA
query = """
SELECT
    v.venta_id,
    e.nombre AS vendedor,
    v.fecha_venta,
    v.cantidad * v.precio_unit AS monto,
    SUM(v.cantidad * v.precio_unit) OVER (
        PARTITION BY e.nombre
        ORDER BY v.fecha_venta
    ) AS acumulado
FROM ventas v
INNER JOIN empleados e ON e.empleado_id = v.empleado_id
"""
pd.read_sql_query(query, conn)
```

```python
#AVG - OVER
query = """
SELECT
    v.venta_id,
    v.categoria,
    v.cantidad * v.precio_unit AS monto,
    ROUND(AVG(v.cantidad * v.precio_unit) OVER (PARTITION BY v.categoria), 2) AS promedio_categoria
FROM ventas v
"""
pd.read_sql_query(query, conn)
```

# Punto 3

Como ejemplo para probar las operaciones de conjunto voy a tomar los empleados que vendieron en el 1er semestre y 2do semestre de 2024

```python
pd.read_sql_query("SELECT * FROM ventas", conn)
```

```python
# UNION, empleados que vendieron en el 1er semestre O en el 2do, sin repetir
query = """
SELECT empleado_id FROM ventas WHERE fecha_venta < '2024-07-01'
UNION
SELECT empleado_id FROM ventas WHERE fecha_venta >= '2024-07-01'
"""
pd.read_sql_query(query, conn)
```

```python
# UNION ALL, lo mismo que pasa con UNION pero con duplicados (el que vendió en ambos semestres aparece 2 veces)
query = """
SELECT empleado_id FROM ventas WHERE fecha_venta < '2024-07-01'
UNION ALL
SELECT empleado_id FROM ventas WHERE fecha_venta >= '2024-07-01'
"""
pd.read_sql_query(query, conn)
```

```python
#INTERSECT, muestra los  empleados que vendieron en AMBOS semestres
query = """
SELECT empleado_id FROM ventas WHERE fecha_venta < '2024-07-01'
INTERSECT
SELECT empleado_id FROM ventas WHERE fecha_venta >= '2024-07-01'
"""
pd.read_sql_query(query, conn)
```

```python
#EXCEPT, muestra los empleados que vendieron en el 1er semestre pero NO en el 2do
query = """
SELECT empleado_id FROM ventas WHERE fecha_venta < '2024-07-01'
EXCEPT
SELECT empleado_id FROM ventas WHERE fecha_venta >= '2024-07-01'
"""
pd.read_sql_query(query, conn)
```

```python
# Usuarios y registros recientes. Nota: si se usaba como referencia la fecha actual (2026) la consulta devolvia datos vacíos, por lo que se toma como referencia el año 2025.
query = """
SELECT
    u.name,
    u.email,
    p.phone,
    p.address,
    u.created_at
FROM users u
INNER JOIN profiles p ON p.user_id = u.user_id
WHERE u.age > 30
  AND strftime('%Y', u.created_at) = '2025'
ORDER BY u.created_at ASC
LIMIT 5
"""
pd.read_sql_query(query, conn2)
```

```python
# Gasto acumulado por cliente
query = """
SELECT
    u.user_id,
    u.name,
    SUM(o.amount) AS gasto_total
FROM users u
INNER JOIN orders o ON o.user_id = u.user_id
GROUP BY u.user_id, u.name
HAVING SUM(o.amount) > 700
ORDER BY gasto_total DESC
"""
pd.read_sql_query(query, conn2)
```

```python
# Ciudades activas
query = """
    SELECT p.address
    FROM profiles p
    INNER JOIN orders o ON o.user_id = p.user_id
    GROUP BY p.user_id, p.address
    HAVING COUNT(o.order_id) >= 2
"""
pd.read_sql_query(query, conn2)
```

```python
#Segmentación demográfica
query = """
SELECT
    CASE
        WHEN age < 25 THEN 'Joven'
        WHEN age BETWEEN 25 AND 40 THEN 'Adulto'
        ELSE 'Senior'
    END AS grupo_edad,
    COUNT(*) AS numero_usuarios
FROM users
GROUP BY grupo_edad
"""
pd.read_sql_query(query, conn2)
```

```python
#Informe combinado de usuarios y pedidos
query = """
SELECT
    u.user_id,
    u.name,
    u.email,
    u.age,
    p.address,
    p.phone,
    o.order_id,
    o.product,
    o.amount,
    o.order_date
FROM users u
INNER JOIN profiles p ON p.user_id = u.user_id
INNER JOIN orders o ON o.user_id = u.user_id
ORDER BY o.order_date DESC
"""

pd.read_sql_query(query, conn2)
```

```python
#1NF todos los datos son atómicos y 2NF
import sqlite3
import pandas as pd
df = pd.read_csv(r'C:\Users\HP\Desktop\Fund_Data_Analysis\Taller_SQL_1\sales.csv', sep=';', encoding='latin1')

print("Tabla original sales.csv:")
df
df
```

```python
conn3 = sqlite3.connect(':memory:')
cur = conn3.cursor()
cur.executescript("""
CREATE TABLE customers (
    customer_id   VARCHAR(10) PRIMARY KEY,
    customer_name VARCHAR(100) NOT NULL,
    customer_city VARCHAR(60)
);

CREATE TABLE products (
    product_id   VARCHAR(10) PRIMARY KEY,
    product_name VARCHAR(100) NOT NULL,
    category     VARCHAR(50)
);

CREATE TABLE reps (
    rep_id   VARCHAR(10) PRIMARY KEY,
    rep_name VARCHAR(100) NOT NULL,
    region   VARCHAR(50)
);


CREATE TABLE sale_items (
    sale_id    INTEGER NOT NULL REFERENCES sales(sale_id),
    product_id VARCHAR(10) NOT NULL REFERENCES products(product_id),
    quantity   INTEGER NOT NULL CHECK (quantity > 0),
    unit_price DECIMAL(12,2) NOT NULL CHECK (unit_price >= 0),
    PRIMARY KEY (sale_id, product_id)
);


CREATE TABLE sales (
    sale_id     INTEGER PRIMARY KEY,
    sale_date   DATE NOT NULL,
    customer_id VARCHAR(10) NOT NULL REFERENCES customers(customer_id),
    rep_id      VARCHAR(10) NOT NULL REFERENCES reps(rep_id)
);

""")

customers  = df[['customer_id','customer_name','customer_city']].drop_duplicates().sort_values('customer_id')
products   = df[['product_id','product_name','category']].drop_duplicates().sort_values('product_id')
reps       = df[['rep_id','rep_name','region']].drop_duplicates().sort_values('rep_id')
sales      = df[['sale_id','sale_date','customer_id','rep_id']].drop_duplicates().sort_values('sale_id')
sale_items = df[['sale_id','product_id','quantity','unit_price']].sort_values('sale_id')

customers.to_sql('customers', conn3, if_exists='append', index=False)
products.to_sql('products', conn3, if_exists='append', index=False)
reps.to_sql('reps', conn3, if_exists='append', index=False)
sales.to_sql('sales', conn3, if_exists='append', index=False)
sale_items.to_sql('sale_items', conn3, if_exists='append', index=False)

conn3.commit()
```

```python
cur.executescript("""
CREATE TABLE dim_customer (
    customer_id   VARCHAR(10) PRIMARY KEY,
    customer_name VARCHAR(100),
    customer_city VARCHAR(60)
);
 
CREATE TABLE dim_product (
    product_id   VARCHAR(10) PRIMARY KEY,
    product_name VARCHAR(100),
    category     VARCHAR(50)
);
 
CREATE TABLE dim_reps (
    rep_id   VARCHAR(10) PRIMARY KEY,
    rep_name VARCHAR(100),
    region   VARCHAR(50)
);
 
CREATE TABLE dim_date (
    date_key  INTEGER PRIMARY KEY,
    full_date DATE
);
 
CREATE TABLE fact_sales (
    sale_id      INTEGER NOT NULL,
    product_id   VARCHAR(10) NOT NULL REFERENCES dim_product(product_id),
    date_key     INTEGER REFERENCES dim_date(date_key),
    customer_id  VARCHAR(10) REFERENCES dim_customer(customer_id),
    rep_id       VARCHAR(10) REFERENCES dim_rep(rep_id),
    quantity     INTEGER,
    unit_price   DECIMAL(12,2),
    total_amount DECIMAL(12,2), 
    PRIMARY KEY (sale_id, product_id)
);

""")
conn3.commit()
```

```python
cur.execute("""
INSERT INTO dim_date (date_key, full_date)
SELECT DISTINCT
    CAST(strftime('%Y%m%d', sale_date) AS INTEGER),
    sale_date
FROM sales;
""")
conn3.commit()

pd.read_sql_query("SELECT * FROM dim_date ORDER BY date_key", conn3)
```
