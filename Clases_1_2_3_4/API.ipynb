# **APIs**

```python
import requests
import time
#Realizar una petición
url = 'https://rickandmortyapi.com/api/character'
response = requests.get(url)
response
```

```python
id_character = 2
response = requests.get(f"{url}/{id_character}")
status_code = response.status_code
if status_code  == 200:
    character = response.json()
    print(character)
```

```python
list_characters = []
next_url = url
while next_url :
    response = requests.get(next_url)
    characters = response.json()
    list_characters.extend(characters['results'])
    print(f"Current url: {next_url}")
    print(f"Number of characters: {len(list_characters)}")
    next_url = characters['info']['next']
    time.sleep(1)
len(list_characters)
    
```

```python
import pandas as pd
data = pd.DataFrame(list_characters)
data.head()
data.gender.value_counts()
```
