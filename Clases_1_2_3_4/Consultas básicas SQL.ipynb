```python
# Cual es la cantidad total, minima, maxima y promedio de articulos vendidos por el estado actual
# de la compra
query = """
SELECT
        estado,
        SUM(cantidad) AS total_cantidad,
        MIN(cantidad) AS min_cantidad,
        MAX(cantidad) AS max_cantidad,
        AVG(cantidad) AS avg_cantidad,
        COUNT(cantidad) AS count_cantidad
FROM pedidos
GROUP BY estado
"""
# Cuando se emplea el comando GROUP BY se debe cumplir con lo siguiente:
# - Las columnas a la derecha GROUP BY son las columnas que deben ir después de SELECT
# - En el SELECT, después de las columnas que aparecen en el GROUP BY, se colocan funciones de agregación: SUM, MIN, MAX, AVG, COUNT
pd.read_sql_query(query, conn)
```

```python
# Cual es el precio unitario minimo, maximo y promedio de articulos vendidos por el estado actual
# de la compra
query = """
SELECT
    estado,
    MIN(precio_unit) AS min_precio_unit,
    MAX(precio_unit) AS max_precio_unit,
    ROUND(AVG(precio_unit),2) AS avg_precio_unit
FROM pedidos
GROUP BY estado
"""
pd.read_sql_query(query, conn)
```

```python
# Cual es el estado que presenta una cantidad de productos total mayor a 5
query = """
SELECT
        estado,
        SUM(cantidad) AS total_cantidad
FROM pedidos
GROUP BY estado
HAVING total_cantidad > 5
"""
# HAVING: filtro que se aplica a los resultados de los GROUP BY
pd.read_sql_query(query, conn)
```

```python
# Muestre la tabla de pedidos pero creando la columna categoria_valor, donde tiene que
# devolver ALTO en el caso en el que el precio unitario supere 1000, si está entre 200 y 1000 debe ser NORMAL
# y para todo caso contrario, colocar BAJO.
query = """
SELECT
  precio_unit,
  CASE
    WHEN precio_unit > 1000
      THEN 'ALTO'
    WHEN precio_unit < 1000 AND precio_unit >= 200
      THEN 'NORMAL'
  ELSE 'BAJO'
  END AS categoria_valor 
FROM
  pedidos
"""
# si no se pone el nombre de la columna (categoria_valor) el código pone un nombre largo raro.
pd.read_sql_query(query, conn)
```

¿Qué pedidos con estado “ENVIADO” o “ENTREGADO” tienen más de una unidad y fueron realizados entre el 1 de enero y el 30 de junio de 2025?

```python
query = """
SELECT
  *
FROM pedidos
WHERE estado IN ('ENVIADO','ENTREGADO')
AND cantidad > 1 AND fecha_pedido BETWEEN '2025-01-01' AND '2025-06-30'
"""
pd.read_sql_query(query, conn)
```

¿Qué clientes han hecho más de un pedido, y cuál es el número de pedidos y el total gastado por cada uno?

```python
query = """
SELECT
  A.cliente_id,
  COUNT(A.cliente_id) AS numero_pedidos,
  SUM(cantidad*precio_unit) AS total_gastado
FROM clientes A
INNER JOIN pedidos B
  ON A.cliente_id = B.cliente_id
GROUP BY A.cliente_id
HAVING numero_pedidos > 1
"""
pd.read_sql_query(query, conn)
```

¿Cómo se clasifica el nivel de gasto de cada cliente en “ALTO”, “MEDIO” o “BAJO”, según el total que ha gastado?

```python
query = """
SELECT
  cliente_id,
  SUM(cantidad*precio_unit) AS total
FROM
  pedidos
GROUP BY cliente_id
ORDER BY total DESC
"""
pd.read_sql_query(query, conn)
```

```python
query = """
WITH total_clientes AS
(
SELECT
  cliente_id,
  SUM(cantidad*precio_unit) AS total
FROM
  pedidos
GROUP BY cliente_id
)
SELECT
  cliente_id,
  total,
  CASE
    WHEN total >= 1000
      THEN 'ALTO'
    WHEN total > 400
      THEN 'MEDIO'
    ELSE 'BAJO'
  END AS categoria
FROM total_clientes
"""

# With me segmenta o me crea sub tablas para ejecutar como en pequeños pasos, para hacer subconsultas
pd.read_sql_query(query, conn)
```

¿Cuáles son las ciudades únicas de los clientes (no nulas), ordenadas alfabéticamente? Muestre los resultados(filas) 2,3 y 4, es decir, muestre tres resultados a partir del segundo.

```python
query = """
SELECT DISTINCT
  ciudad
FROM clientes
WHERE ciudad IS NOT NULL
ORDER BY ciudad ASC
LIMIT 3
OFFSET 1
"""

# DISTINCT borra las variables repetidas
# LIMIT me limita las filas que puedo ver, solo me deja ver X cantidad de cosas
# OFFSET ejecutar el limite apartir de una fila X y no desde la primera fila

pd.read_sql_query(query, conn)
```
