## 3. Comprensión del dataset

```python
print(f"La base de datos presenta las siguientes dimnensiones: {data.shape}")
data.info()
```

**¿Qué información contiene el dataset?**

Cada fila del dataset es una posición GPS registrada por hora para un oso individual (de 6 poblaciones de osos de 11 países), junto con variables de comportamiento (estado de movimiento, período del día) y variables ambientales asociadas a ese momento y lugar (temperatura máxima, duración del día, altitud solar, índice de vegetación NDVI).

**Columnas del dataset:**
- `id`: identificador del oso.
- `Sex`: sexo codificado en binario donde, 1=female, 2=male.
- `age`: edad del oso codificado de la siguiente manera, 1=subadult (3/4 years old), 2=adult (>= 5 years old).
- `movem_60`: velocidad de movimiento en la última hora (Km/h).
- `timestamp`: fecha y hora UTC del registro.
- `season`: estación codificada entre 1 a 3, 1=spring, 2=summer, 3=fall.
- `reprod`: estado reproductivo codificado entre 1 a 3, 1=lone female, 2=male, 3=female accompanied by young
- `move_status`: estado de movimiento categórico si el oso es residente o migrante.
- `cor_hour`: hora del día corregida (1–24h).
- `altitude_dg`: altitud solar en grados (negativa de noche, positiva de día).
- `t_period`: período del día — `nocturnal`, `crepuscular`, `diurnal`.
- `index`: un índice numérico asociado al registro (probablemente de calidad de la señal GPS).
- `tmx`: temperatura máxima del día (°C).
- `dayle`: duración del día (horas de luz, "day length").
- `study`, `year`: identificadores del estudio y año.
- `ndvi`: índice de vegetación (NDVI), proxy de disponibilidad de alimento/vegetación.

```python
print('Registros y variables:', data.shape[0], 'filas x', data.shape[1], 'columnas')
```

**Variables más importantes para el análisis:** 
- `movem_60` la velocidad de desplazamiento del oso, que es la variable más importante del dataset de esta se podria decir que se usa para el analissi del repositorio.
-  `timestamp` / `cor_hour` son importantes para ver las horas en las que el oso se desplazó
- `t_period` la categoría en la que se desplaza el oso, podriamos evidenciar si el oso tiene más desplazamiento nocturno o diurno.
- `tmx`, `ndvi`, `dayle` y `altitude_dg` (variables ambientales que podrían explicar el movimiento).
- `season` para comparar temporadas.

```python
# Valores nulos
print('Nulos por columna:')
print(data.isnull().sum())
print()
# Duplicados
print('Filas duplicadas:', data.duplicated().sum())
```

```python
(data.isnull().sum()/data.shape[0]).round(3)*100
```

**El dataset no presenta duplicados, sin embargo, hay 2 variables que contienen nulos en sus registros:**
- `Age` (edad) = presenta 1329 datos nulos, que representa el 0.4% de los registros totales
- `Reprod` (estado reproductivo) = presenta 50300 datos nulos, que representa el 13,2% de los registros.

```python
variables_interes = ['movem_60', 'tmx', 'ndvi', 'dayle', 'altitude_dg', 'season']

for col in variables_interes:
    q1, q3 = data[col].quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5*iqr, q3 + 1.5*iqr
    outliers = data[(data[col] < lower) | (data[col] > upper)]

    print(f'--- {col} ---')
    print(f'Rango intercuartílico normal: [{lower:.3f}, {upper:.3f}]')
    print(f'Registros atípicos: {len(outliers)} de {len(data)} ({len(outliers)/len(data):.1%})')
    display(outliers[[col]].head())
    print()
```

**De los variables importante de análisis se encontraron que 4 de 6 presentan datos atipicos distribuidos de la siguiente forma:**
- `movem_60`: contiene 42084 de 379676 datos son atipicos que representa el 11.1% de los regristros
- `tmx`: contiene 1674 de 379676 datos son atipicos que representa el 0.4%% de los regristros
- `ndvi`: contiene 15940 de 379676 datos son atipicos que representa el 4.2% de los regristros
- `dayle`: contiene 5609 de 379676 datos son atipicos que representa el 1.5% de los regristros
